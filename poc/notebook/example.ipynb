# Example Event Extraction

This will do some simple event extraction examples.



```python
from __future__ import annotations

import pandas as pd
from ollama import Client

from poc.prompt import et_prompt
from poc.query import get_ollama_responce
```

```python
data_df = pd.read_json(
    "../data/gpt_data_with_embeddings.json", orient="records", lines=True
)
```

```python
from numpy.random import RandomState

random_state = RandomState(42)

samples = random_state.choice(data_df.index, 50, replace=False)
```

```python
samples
```

```python
some_examples = [data_df.loc[x].to_dict() for x in [0, 23, 45, 21, 102]]
```

```python
et_prompt("Jane went to the store to buy some milk.", examples=some_examples)
```

```python
ollama_client = Client()
```

```python
get_ollama_responce(
    ollama_client,
    "llama3.2:3b",
    et_prompt("Jane went to the store to buy some milk.", examples=some_examples),
)
```
