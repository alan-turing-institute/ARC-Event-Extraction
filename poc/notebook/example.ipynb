# Example Event Extraction

This will do some simple event extraction examples.



```python
from __future__ import annotations

import pandas as pd
from ollama import Client

from poc.prompt import et_prompt
from poc.query import get_ollama_responce, hand_close_examples
```

```python
data_df = pd.read_json(
    "../data/gpt_data_with_embeddings.json", orient="records", lines=True
)
```

```python
from numpy.random import RandomState

random_state = RandomState(42)

sample_choice = random_state.choice(data_df.index, 10, replace=False)
```

```python
sample_choice
```

```python
ollama_client = Client()
```

```python
def query_ollama(sent, examps):
    prompt = et_prompt(sent, examples=examps)
    return get_ollama_responce(ollama_client, "llama3.2:3b", prompt)


random_things = hand_close_examples(
    data_df,
    sample_choice,
    query_ollama,
    verbose=True,
)
```

```python
for samp, res in random_things:
    print(f"Sentence: {samp['Sentence']} , answer: {samp['EventType']}")
    print(f"Response: {res}")
    print("-------------------")
```
