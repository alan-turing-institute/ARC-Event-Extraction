# Trying Inference with Huggingface Transformers

This is just following the guide online for using pipelines.

[Pipelines](https://huggingface.co/docs/transformers/en/main_classes/pipelines)

[Llama Models](https://huggingface.co/meta-llama)


```python
from __future__ import annotations

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

MODEL_LOC = "../model/Llama-3.2-1B-Instruct"
```

```python
torch.random.manual_seed(0)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_LOC,
    device_map="auto",
    torch_dtype="auto",
    trust_remote_code=True,
)
tokenizer = AutoTokenizer.from_pretrained(MODEL_LOC)

messages = [
    {"role": "system", "content": "You are a helpful AI assistant."},
    {
        "role": "user",
        "content": (
            "Can you provide ways to eat combinations of " "bananas and dragonfruits?"
        ),
    },
    {
        "role": "assistant",
        "content": (
            "Sure! Here are some ways to eat bananas and dragonfruits together: "
            "1. Banana and dragonfruit smoothie: Blend bananas and dragonfruits "
            "together with some milk and honey. 2. Banana and dragonfruit salad: "
            "Mix sliced bananas and dragonfruits together with some lemon juice "
            "and honey."
        ),
    },
    {"role": "user", "content": "What about solving an 2x + 3 = 7 equation?"},
]

pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
)

generation_args = {
    "max_new_tokens": 500,
    "return_full_text": False,
    "temperature": 0.0,
    "do_sample": False,
}

output = pipe(messages, **generation_args)
print(output[0]["generated_text"])
```

```python
pipe(
    [
        {"role": "system", "content": "You are a helpful AI assistant."},
        {"role": "user", "content": "What does head of the ECB actually do?"},
    ],
    **generation_args,
)
```

```python
pipe(
    [
        {"role": "system", "content": "You are a helpful AI assistant."},
        {"role": "user", "content": "Who made you?"},
    ],
    **generation_args,
)
```
